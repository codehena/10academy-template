```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

print("Ready! pandas:", pd.__version__)
```

```python
# Example using a small sample DataFrame (replace with pd.read_csv("../data/your_file.csv") when ready)
df = pd.DataFrame({
    "category": ["A", "B", "A", "C", "B", "A"],
    "value": [12, 19, 15, 8, 22, 17]
})

print("Shape:", df.shape)
print(df.info())
print(df.isnull().sum())
df.head()
```

```python
sns.barplot(data=df, x="category", y="value", estimator="mean")
plt.title("Mean Value by Category")
plt.tight_layout()
plt.show()
```
